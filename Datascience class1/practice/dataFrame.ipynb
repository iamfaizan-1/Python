## creating dataframes



### 1. Option 1 (numpy)


```python
import numpy as np
import pandas as pd
```

```python
# creating array
data = np.array([[1,2],[3,4],[5,6],[7,8]])
```

```python
# creating dataframe
df = pd.DataFrame(data,index=["row1","row2","row3","row4"], columns=["col1","col2"])
print(df)
```

### 1.2 Option 2 (Lists)

```python
# creating an array with list

data2 = [[1,2],[3,4],[5,6]] 
```

```python
df = pd.DataFrame(data2,index=["row1","row2","row3"],columns=["col1","col2"])
df
```

### 2. Dictionary


```python
states = ["California","Texas","Florida","New york"]
population=[3242342,345345345,234234,6446456]
```
